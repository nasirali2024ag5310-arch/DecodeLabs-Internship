```python
import pandas as pd 

# loading my project data file
df = pd.read_excel(r"..\data\Dataset for Data Analytics.xlsx")
```

```python
# checking structre of dataset
print(df.info())
```

```python
# checking total missing values across each column
print("Missing Values:\n",df.isnull().sum())
```

```python
# checking initial dataset
print(df.head())
```

```python
"""
handling missing values
blank CouponCode filled with "No CouponCode used"
"""
df["CouponCode"]=df["CouponCode"].fillna("No CouponCode")
```

```python
# check duplicate values
total_duplicates=df.duplicated().sum()
print("Duplicates: " , total_duplicates)
```

```python
# remove duplicates if any
df=df.drop_duplicates()
```

```python
# checking date format
print(pd.to_datetime(df["Date"], errors ="coerce").isna().sum())
```

```python
# Standard date format
df["Date"] = df["Date"].dt.strftime("%Y-%m-%d")
```

```python
# checking date format
print(pd.to_datetime(df["Date"], errors ="coerce").isna().sum())
```

```python
# clean text columns
text_cols=["OrderID","CustomerID","Product","PaymentMethod","ShippingAddress","OrderStatus","TrackingNumber","CouponCode","ReferralSource"]
for col in text_cols:
    df[col]=df[col].str.strip()
```

```python
# clean and round price columns
price_cols = ["UnitPrice", "TotalPrice"]
for col in price_cols:
    df[col] = (df[col].astype(str).str.replace("$", "", regex=False).str.replace(",", "", regex=False)).astype("float64")
    df[col] = df[col].round(2)
```

Validating data after cleaning

```python
# check missing values
df.isnull().sum()
```

```python
# check duplicate rows
print(df.duplicated().sum())
```

```python
# check duplicate IDs
print(df["OrderID"].duplicated().sum())
```

```python
# check date format
print(pd.to_datetime(df["Date"], errors="coerce").isna().sum())
```

```python
# check data types
print(df.dtypes)
```

```python
# check numeric values
print(df.describe())
```

```python
# saving final cleaned data 
df.to_excel(r"..\data\Cleaned_Dataset.xlsx", index=False)
```
